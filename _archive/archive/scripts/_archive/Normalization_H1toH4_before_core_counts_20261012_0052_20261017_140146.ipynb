# Physiological transformations for H1 to H4

One row per physiological column in `C:\Users\pandya\Documents\Github\docker\Paper3_Github\output\merged_all_11participants_8min.csv` (SSOT analysis input).

**No transformation** means the column is modelled as stored. **Natural log** means ln(x) for values above zero; rows at or below zero are dropped. That is the heart-rate variability rule. The other physiological columns are not transformed.

`col_name` is the column in the merged file. `Papers` points at the numbered list under the table. That list gives the file title and the full path.

`Transformation in Step 5` (in `Normalization_H1toH4_physio.csv`) is generated from the same rule as `Transformation`, so the two columns cannot disagree: heart-rate variability columns use ln(x) with rows at or below zero excluded (not ln(1 + x)); this matches `H4_SPEC$outcome_transform` in `scripts/H4_multivariate.R`. Columns not modelled in H4 Step 5 are marked `Not in Step 5`.

Layers: 1 Config, 2 Load, 3 Define, 4 Run, then one Save cell. All outputs are written to `C:\Users\pandya\Documents\Github\docker\Paper3_Final\scripts\`.

```python
# 2. Load
raw = pd.read_csv(DATA, low_memory=False)
missing_cols = [c for c in COL.values() if c not in raw.columns]
assert not missing_cols, missing_cols
```

```python
# 3. Define
def locate(key):
    hits = [p for p in PAPERS.rglob("*" + key + "*") if p.suffix.lower() == ".pdf"]
    if len(hits) != 1:
        raise FileNotFoundError(key + " -> " + str([h.name for h in hits]))
    return hits[0]

def step5_text(col_name, transformation):
    return "Not in Step 5" if col_name in NOT_IN_STEP5 else STEP5_TEXT[transformation]

def check_log_domain(col_name, transformation):
    # Rows excluded by the ln(x) rule (x <= 0); reported so the exclusion is visible.
    x = pd.to_numeric(raw[col_name], errors="coerce")
    return int((x <= 0).sum()) if transformation == "Natural log" else 0
```

```python
# 4. Run
catalog = []
number_of = {}
rows = []
physio_rows = []
for parameter, options in USED.items():
    transformation, keys = options[0]
    marks, titles = [], []
    for key in keys:
        path = locate(key)
        if path.name not in number_of:
            number_of[path.name] = len(catalog) + 1
            catalog.append({"Number": f"[{number_of[path.name]}]", "Paper title": path.name, "File path": str(path)})
        marks.append(f"[{number_of[path.name]}]")
        titles.append(path.name)
    col_name = COL[parameter]
    rows.append({
        "col_name": col_name,
        "Parameter": parameter,
        "Transformation": transformation,
        "Number of papers": len(marks),
        "Papers": ", ".join(marks),
    })
    physio_rows.append({
        "Column": col_name,
        "Parameter": parameter,
        "Unit": UNIT[col_name],
        "Transformation in the papers": transformation,
        "Transformation in Step 5": step5_text(col_name, transformation),
        "Rows excluded (x <= 0)": check_log_domain(col_name, transformation),
        "Paper titles": " | ".join(titles),
    })

count = pd.DataFrame(rows)
papers_table = pd.DataFrame(catalog)
physio = pd.DataFrame(physio_rows)
assert not physio["Transformation in Step 5"].str.contains(r"1 \+ x|log1p", regex=True).any()
pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_rows", 80)
display(count)
display(papers_table)
display(physio)
```

```python
# Save (single output cell)
count.to_csv(OUT_COUNTS, index=False)
papers_table.to_csv(OUT_INDEX, index=False)
physio.to_csv(OUT_PHYSIO, index=False)
print(OUT_COUNTS, OUT_INDEX, OUT_PHYSIO, sep="\n")
```
