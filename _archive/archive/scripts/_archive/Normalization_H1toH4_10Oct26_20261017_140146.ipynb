# Physiological transformations for H1 to H4

One row per physiological column in `output/merged_all_11participants_8min.csv`.

**No transformation** means the column is modelled as stored. **Natural log** means ln(x) for values above zero; rows at or below zero are dropped. That is the heart-rate variability rule. The other physiological columns are not transformed.

`col_name` is the column in the merged file. `Papers` points at the numbered list under the table. That list gives the file title and the full path.


```python
from pathlib import Path
import pandas as pd
from IPython.display import display

ROOT = Path(r"C:/Users/pandya/Documents/Github/docker/Paper3_Final")
PAPERS = Path(r"C:/Users/pandya/Documents/Github/docker/Paper3_Github/Experiment papers")
raw = pd.read_csv(ROOT / "output" / "merged_all_11participants_8min.csv", low_memory=False)

# Parameter, the transformation, and how many papers used that transformation.
# One row per parameter. Counted only when the paper uses that transformation.
# A paper that only rejects another method is not counted.

USED = {
    "Heart rate": [("No transformation", ["Zhang et al 2020", "Beggiato et al 2018", "Outdoor temperature heart rate", "Personal_thermal_comfort_wearable_sensors_2018", "Thermal_comfort_physiological_sensing_DTU_2025", "Tervonen"])],
    "Skin temperature": [("No transformation", ["Zhang et al 2020", "Personal_thermal_comfort_wearable_sensors_2018", "Thermal_comfort_physiological_sensing_DTU_2025", "Kyriakou", "novel methodology for human thermal comfort"])],
    "Skin conductance level": [("No transformation", ["Fowles", "EDA_recommendations_BIOPAC", "Vocks et al 2021", "Zhang et al 2020", "Figner_Murphy_2010"])],
    "Skin conductance, tonic": [("No transformation", ["Fowles", "EDA_recommendations_BIOPAC", "Vocks et al 2021", "Zhang et al 2020", "Figner_Murphy_2010"])],
    "Skin conductance, phasic": [("No transformation", ["Fowles", "EDA_recommendations_BIOPAC", "Vocks et al 2021", "EDA_transform_BIOPAC_EDA_SCR", "Figner_Murphy_2010"])],
    "SDNN": [("Natural log", ["Laborde Mosley Thayer 2017", "An Overview of Heart Rate Variability Metrics and Norms", "HRV_task_force", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "RMSSD": [("Natural log", ["Laborde Mosley Thayer 2017", "An Overview of Heart Rate Variability Metrics and Norms", "HRV_task_force", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "LF power": [("Natural log", ["Laborde Mosley Thayer 2017", "An Overview of Heart Rate Variability Metrics and Norms", "HRV_task_force", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "HF power": [("Natural log", ["Laborde Mosley Thayer 2017", "An Overview of Heart Rate Variability Metrics and Norms", "HRV_task_force", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "VLF power": [("Natural log", ["Laborde Mosley Thayer 2017", "An Overview of Heart Rate Variability Metrics and Norms", "HRV_task_force", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "Total power": [("Natural log", ["Laborde Mosley Thayer 2017", "An Overview of Heart Rate Variability Metrics and Norms", "HRV_task_force", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "LF/HF ratio": [("Natural log", ["Laborde Mosley Thayer 2017", "An Overview of Heart Rate Variability Metrics and Norms", "HRV_task_force", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "LF normalized": [("No transformation", ["HRV_task_force", "An Overview of Heart Rate Variability Metrics and Norms", "Laborde Mosley Thayer 2017", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "HF normalized": [("No transformation", ["HRV_task_force", "An Overview of Heart Rate Variability Metrics and Norms", "Laborde Mosley Thayer 2017", "Reliability of Ultra-Short-Term", "Threshold-Based Artefact Correction"])],
    "Pupil diameter": [("No transformation", ["Imburgio et al 2024", "Beggiato et al 2018", "Mahanama et al 2021", "Wang et al 2025", "Wang et al 2024"])],
    "Pupil change rate": [("No transformation", ["Imburgio et al 2024", "Beggiato et al 2018", "Mahanama et al 2021", "Wang et al 2025", "Wang et al 2024"])],
    "Eyelid aperture": [("No transformation", ["Mahanama et al 2021", "Beggiato et al 2018", "Wang et al 2025", "Wang et al 2024", "Evans et al 2012"])],
    "Blink duration": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Rupi and Krizek 2019", "Trefzger et al 2018", "Vansteenkiste et al 2014"])],
    "Fixation duration": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Rupi and Krizek 2019", "Trefzger et al 2018", "Vansteenkiste et al 2014"])],
    "Saccade duration": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Rupi and Krizek 2019", "Trefzger et al 2018", "Vansteenkiste et al 2014"])],
    "Fixation rate": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Rupi and Krizek 2019", "Trefzger et al 2018", "Vansteenkiste et al 2014"])],
    "Saccade amplitude": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Rupi and Krizek 2019", "Trefzger et al 2018", "Vansteenkiste et al 2014"])],
    "Saccade peak velocity": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Rupi and Krizek 2019", "Trefzger et al 2018", "Vansteenkiste et al 2014"])],
    "Gaze velocity": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Rupi and Krizek 2019", "Trefzger et al 2018", "Vansteenkiste et al 2014"])],
    "Gaze dispersion": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Rupi and Krizek 2019", "Trefzger et al 2018", "Vansteenkiste et al 2014"])],
    "Distance from center": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Fleming et al 2024"])],
    "Gaze centrality": [("No transformation", ["Wang et al 2025", "Wang et al 2024", "Evans et al 2012", "Tomasi et al 2016", "Fleming et al 2024"])],
}

COL = {
    "Heart rate": "empatica_heart_rate",
    "Skin temperature": "empatica_skin_temperature",
    "Skin conductance level": "empatica__eda_scl_usiemens",
    "Skin conductance, tonic": "empatica_eda_tonic",
    "Skin conductance, phasic": "empatica_eda_phasic",
    "SDNN": "empatica_hrv_td_sdnn",
    "RMSSD": "empatica_hrv_td_rmssd",
    "LF power": "empatica_hrv_fd_lf",
    "HF power": "empatica_hrv_fd_hf",
    "VLF power": "empatica_hrv_fd_vlf",
    "Total power": "empatica_hrv_fd_total_power",
    "LF/HF ratio": "empatica_hrv_fd_lf_hf_ratio",
    "LF normalized": "empatica_hrv_fd_lfnu",
    "HF normalized": "empatica_hrv_fd_hfnu",
    "Pupil diameter": "eyetracker_pupil_diameter_avg",
    "Pupil change rate": "eyetracker_pupil_change_rate",
    "Eyelid aperture": "eyetracker_eyelid_aperture_avg",
    "Blink duration": "eyetracker_blink_duration_s",
    "Fixation duration": "eyetracker_fixation_duration_s",
    "Saccade duration": "eyetracker_saccade_duration_s",
    "Fixation rate": "eyetracker_fixation_rate",
    "Saccade amplitude": "eyetracker_saccade_amplitude",
    "Saccade peak velocity": "eyetracker_saccade_peak_velocity",
    "Gaze velocity": "eyetracker_gaze_velocity",
    "Gaze dispersion": "eyetracker_gaze_dispersion",
    "Distance from center": "eyetracker_distance_from_center",
    "Gaze centrality": "eyetracker_gaze_centrality",
}
missing_cols = [c for c in COL.values() if c not in raw.columns]
assert not missing_cols, missing_cols

def locate(key):
    hits = [p for p in PAPERS.rglob("*" + key + "*") if p.suffix.lower() == ".pdf"]
    if len(hits) != 1:
        raise FileNotFoundError(key + " -> " + str([h.name for h in hits]))
    return hits[0]

catalog = []
number_of = {}
rows = []
for parameter, options in USED.items():
    transformation, keys = options[0]
    marks = []
    for key in keys:
        path = locate(key)
        if path.name not in number_of:
            number_of[path.name] = len(catalog) + 1
            catalog.append({"Number": f"[{number_of[path.name]}]", "Paper title": path.name, "File path": str(path)})
        marks.append(f"[{number_of[path.name]}]")
    rows.append({
        "col_name": COL[parameter],
        "Parameter": parameter,
        "Transformation": transformation,
        "Number of papers": len(marks),
        "Papers": ", ".join(marks),
    })

count = pd.DataFrame(rows)
papers_table = pd.DataFrame(catalog)
count.to_csv(ROOT / "scripts" / "Normalization_H1toH4_counts.csv", index=False)
papers_table.to_csv(ROOT / "scripts" / "Normalization_H1toH4_paper_index.csv", index=False)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_rows", 80)
display(count)
display(papers_table)
```
